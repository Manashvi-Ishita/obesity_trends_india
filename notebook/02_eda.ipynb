### Step 7: Zone Wise Average Analysis

```python
zone_obesity = df.groupby('zone')[['overweight_obese_w', 'overweight_obese_m']].mean().round(1)
print(zone_obesity.sort_values('overweight_obese_w', ascending=False))
```

### Step 8: Top 5 and Bottom 5 States by Obesity

```python
sorted_w = df[['state','overweight_obese_w']].sort_values('overweight_obese_w', ascending=False)

print("=====Women=====")
print("\nThe top 5 states are:",sorted_w.head())
print("\nThe bottom 5 states are:",sorted_w.tail())
```

```python
sorted_m = df[['state','overweight_obese_m']].sort_values('overweight_obese_m', ascending=False)

print("=====Men=====")
print("\nThe top 5 states are:",sorted_m.head())
print("\nThe bottom 5 states are:",sorted_m.tail())
```

### Step 9: Correlation Analysis
**Note:** Correlation measures how two variables move together.
- Values close to +1 = strong positive relationship
- Values close to -1 = strong negative relationship
- Values close to 0 = no relationship

### Step 9.1: Correlation - Obesity vs Education

```python
corr_edu = df[['overweight_obese_w', 'overweight_obese_m','education_w', 'education_m']].corr().round(2)
print(corr_edu)
```

### Step 9.2: Correlation - Obesity vs Income

```python
corr_income = df[['overweight_obese_w','overweight_obese_m','income']].corr().round(2)
print(corr_income)
```

### step 9.3: Correlation - Obesity vs Tobacco

```python
corr_tobacco = df[['overweight_obese_w','overweight_obese_m','tobacco_w','tobacco_m']].corr().round(2)
print(corr_tobacco)
```

### Step 9.4: Correlation - Obesity vs Alcohol

```python
corr_alcohol = df[['overweight_obese_w','overweight_obese_m','alcohol_w','alcohol_m']].corr().round(2)
print(corr_alcohol)
```

### Step 9.5: Correlation - Obesity vs Internet

```python
corr_internet = df[['overweight_obese_w', 'overweight_obese_m','internet_w', 'internet_m']].corr().round(2)
print(corr_internet)
```

### Step 9.6: Correlation - Obesity vs Blood Sugar

```python
corr_sugar = df[['overweight_obese_w', 'overweight_obese_m','blood_sugar_w', 'blood_sugar_m']].corr().round(2)
print(corr_sugar)
```

### Step 9.7: Correlation - Obesity vs Blood Pressure

```python
corr_bp = df[['overweight_obese_w', 'overweight_obese_m','bp_w', 'bp_m']].corr().round(2)
print(corr_bp)
```
