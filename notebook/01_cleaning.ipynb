### Step 1: Importing Important Libraries

```python
import pandas as pd
import numpy as np
print("Libraries Imported Successfully")
```

### Step 2: Loading the Dataset 
**Source:** NFHS-5 (2019-20), Ministry of Health and Family Welfare, Government of India

```python
df = pd.read_excel('../data/raw/Obesity_Analysis.xlsx', na_values=['NA', 'N/A', '', ' '])
print("Data Loaded Successfully")
```

### Step 3: Checking the Shape of the Dataset

```python
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

### Step 4: Preview the Dataset

```python
df.head()
```

### Step 5: Checking Data Types

```python
df.dtypes
```

### Step 6: Checking Missing Values

```python
df.isnull().sum()
```

### Step 7: Identifying Missing Income States

```python
df[df['income(in Rs.)'].isnull()][['state','regions']]
```

### Step 8: Fixing Income Data Type

```python
df['income(in Rs.)'] = df['income(in Rs.)'].astype('Int64')

print("Done!")
df.head()
```

### Step 9: Strip any extra spaces from column names

```python
df.columns = df.columns.str.strip()
print("Successfull")
```

### Step 10: Renaming Columns

```python
df.rename(columns={
    'income(in Rs.)' : 'income',
    'regions'        : 'zone'
}, inplace=True)

print(df.columns.tolist())
```

### Step 11: Save cleaned dataset

```python
df.to_csv('../data/processed/obesity_cleaned.csv',index=False)
print("Saved Successfully!")
```
