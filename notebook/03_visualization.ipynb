# Obesity Trends in India
## NFHS-5 (2019-20)
### Step 3: Data Visualization

### Step 1: Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print("Library Imported Successfully!")
```

### Step 2: Loading Clean Dataset

```python
df = pd.read_csv('../data/processed/obesity_cleaned.csv')
print("Dataset Loaded Successfully!")
```

### Step 3: Checking Shape 

```python
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

### Step 4: Top 10 States by Obesity - Women vs Men

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Women
top10_w = df.nlargest(10, 'overweight_obese_w')
sns.barplot(data=top10_w, x='overweight_obese_w', y='state',color='coral', ax=axes[0])
axes[0].set_title('Top 10 States - Women')
axes[0].set_xlabel('Obesity (%)')
axes[0].set_ylabel('State/UT')

# Men
top10_m = df.nlargest(10, 'overweight_obese_m')
sns.barplot(data=top10_m, x='overweight_obese_m', y='state', color='steelblue', ax=axes[1])
axes[1].set_title('Top 10 States - Men')
axes[1].set_xlabel('Obesity (%)')
axes[1].set_ylabel('State/UT')

plt.suptitle('Top 10 States by Obesity - NFHS-5 (2019-20)', fontsize=14)
plt.tight_layout()
plt.savefig('../images/top10_obesity.png', dpi=150, bbox_inches='tight')
plt.show()
plt.show()
```

### Step 5: Zone-wise Average Obesity

```python
zone_avg = df.groupby('zone')[['overweight_obese_w', 'overweight_obese_m']].mean()
zone_avg = zone_avg.round(1)
zone_avg = zone_avg.reset_index()

zone_w = zone_avg.sort_values('overweight_obese_w', ascending=False)
zone_m = zone_avg.sort_values('overweight_obese_m', ascending=False)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(data=zone_w, x='zone', y='overweight_obese_w', color='coral', ax=axes[0])
axes[0].set_title('Zone-wise Obesity - Women')
axes[0].set_xlabel('Zone')
axes[0].set_ylabel('Obesity (%)')
axes[0].tick_params(axis='x', rotation=45)

sns.barplot(data=zone_m, x='zone', y='overweight_obese_m', color='steelblue', ax=axes[1])
axes[1].set_title('Zone-wise Obesity - Men')
axes[1].set_xlabel('Zone')
axes[1].set_ylabel('')
axes[1].tick_params(axis='x', rotation=45)

plt.suptitle('Zone-wise Average Obesity - NFHS-5 (2019-20)', fontsize=14)
plt.tight_layout()
plt.savefig('../images/zone_obesity.png', dpi=150, bbox_inches='tight')
plt.show()
```

### Step 6: Obesity vs Income

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Women
sns.scatterplot(data=df, x='income', y='overweight_obese_w', color='coral', ax=axes[0])
axes[0].set_title('Obesity vs Income - Women')
axes[0].set_xlabel('Per Capita Income (Rs.)')
axes[0].set_ylabel('Obesity (%)')

# Men
sns.scatterplot(data=df, x='income', y='overweight_obese_m', color='steelblue', ax=axes[1])
axes[1].set_title('Obesity vs Income - Men')
axes[1].set_xlabel('Per Capita Income (Rs.)')
axes[1].set_ylabel('')

plt.suptitle('Obesity vs Per Capita Income - NFHS-5 (2019-20)', fontsize=14)
plt.subplots_adjust(wspace=0.3)
plt.savefig('../images/obesity_vs_income.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Women
sns.scatterplot(data=df, x='education_w', y='overweight_obese_w', color='coral', ax=axes[0])
axes[0].set_title('Obesity vs Education - Women')
axes[0].set_xlabel('Women with 10+ Years Schooling (%)')
axes[0].set_ylabel('Obesity (%)')

# Men
sns.scatterplot(data=df, x='education_m', y='overweight_obese_m', color='steelblue', ax=axes[1])
axes[1].set_title('Obesity vs Education - Men')
axes[1].set_xlabel('Men with 10+ Years Schooling (%)')
axes[1].set_ylabel('Obesity (%)')

plt.suptitle('Obesity vs Education - NFHS-5 (2019-20)', fontsize=14)
plt.subplots_adjust(wspace=0.3)
plt.savefig('../images/obesity_vs_education.png', dpi=150, bbox_inches='tight')
plt.show()
```

### Step 8: Scatter Plot - Obesity vs Internet Usage

```python
fig, axes = plt.subplots(1,2, figsize=(14,5))

# Women
sns.scatterplot(data=df,x='internet_w',y='overweight_obese_w',color='coral',ax=axes[0])
axes[0].set_title('Obesity vs Internet - Women')
axes[0].set_xlabel('Women who use Internet (%)')
axes[0].set_ylabel('Obesity (%)')

# Men
sns.scatterplot(data=df,x='internet_w',y='overweight_obese_m',color='steelblue',ax=axes[1])
axes[1].set_title('Obesity vs Internet - Men')
axes[1].set_xlabel('Men who use Internet (%)')
axes[1].set_ylabel('Obesity (%)')

plt.suptitle('Obesity vs Internet Usage - NFHS-5 (2019-20)', fontsize=14)
plt.subplots_adjust(wspace=0.3)
plt.savefig('../images/obesity_vs_internet.png', dpi=150, bbox_inches='tight')
plt.show()
```

### Step 9: Scatter Plot - Obesity vs Tobacco

```python
fig, axes = plt.subplots(1,2, figsize=(14,5))

# Women
sns.scatterplot(data=df,x='tobacco_w',y='overweight_obese_w',color='coral',ax=axes[0])
axes[0].set_title('Obesity vs Tobacco - Women')
axes[0].set_xlabel('Women who use Tobacco (%)')
axes[0].set_ylabel('Obesity (%)')

# Men
sns.scatterplot(data=df,x='tobacco_m',y='overweight_obese_m',color='steelblue',ax=axes[1])
axes[1].set_title('Obesity vs Tobacco - Men')
axes[1].set_xlabel('Men who use Tobacco (%)')
axes[1].set_ylabel('Obesity (%)')

plt.suptitle('Obesity vs Tobacco Usage - NFHS-5 (2019-20)', fontsize=14)
plt.subplots_adjust(wspace=0.3)
plt.savefig('../images/obesity_vs_tobacco.png', dpi=150, bbox_inches='tight')
plt.show()
```

### Step 10: Correlation Heatmap

```python
corr_cols = ['overweight_obese_w', 'overweight_obese_m',
             'education_w', 'education_m',
             'income', 'internet_w', 'internet_m',
             'tobacco_w', 'tobacco_m',
             'alcohol_w', 'alcohol_m',
             'blood_sugar_w', 'blood_sugar_m',
             'bp_w', 'bp_m']

corr_matrix = df[corr_cols].corr().round(2)

plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', 
            fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap - NFHS-5 (2019-20)', fontsize=14)
plt.tight_layout()
plt.savefig('../images/correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```
